### Associate Street Network Metrics

```python
node_density_km = []
street_density_km = []
circuity_avg = []
street_length_total = []
street_segment_count = []
streets_per_node_avg = []
orientation_entropy = []
street_statefps = []


# county_networks = []

state_geoids = ['08', '22', '36']

filtered_gpd_eval_results_vis = gpd_eval_results_vis[gpd_eval_results_vis['STATEFP'].isin(state_geoids)]
print(filtered_gpd_eval_results_vis.shape)

areas = filtered_gpd_eval_results_vis.to_crs(epsg = 5070).area

for idx, row in filtered_gpd_eval_results_vis.to_crs(epsg = 4326).iterrows():
    try:
        G = ox.graph.graph_from_polygon(row['geometry'], network_type='all')
        G = ox.convert.to_undirected(G)
        stats = ox.stats.basic_stats(G, area = areas[idx])
        node_density_km.append(stats['node_density_km'])
        street_density_km.append(stats['street_density_km'])
        circuity_avg.append(stats['street_density_km'])
        street_length_total.append(stats['street_length_total'])
        street_segment_count.append(stats['street_segment_count'])
        streets_per_node_avg.append(stats['streets_per_node_avg'])

        G = ox.bearing.add_edge_bearings(G)
        oe = ox.bearing.orientation_entropy(G)
        orientation_entropy.append(oe)
    except Exception as e:
        print(e)
        node_density_km.append(None)
        street_density_km.append(None)
        circuity_avg.append(None)
        street_length_total.append(None)
        street_segment_count.append(None)
        streets_per_node_avg.append(None)
        orientation_entropy.append(None)
```

### Spatial error model

```python
from spreg import ML_Error, GM_Combo, gets_sdm


def partial_r2(model, y, X, w):
    ssr_full = np.sum(model.u**2)
    results = {}

    for j, name in enumerate(X):
        # remove predictor j
        X_reduced = np.delete(X, j, axis=1)
        # m_reduced = ML_Error(y, X_reduced, w=w)
        m_reduced  = gets_sdm(y, X_reduced, w=W)[1]
        ssr_reduced = np.sum(m_reduced.u**2)

        partial_r2_j = 1-  ssr_full / ssr_reduced
        results[name] = partial_r2_j

    return pd.Series(results)


X = df_analysis_per[svi_measures]
# y = gdf_analysis_per['CPC'].values.reshape((-1,1))
y = np.log(df_analysis_per['RMSE'] + 1)

model = ML_Error(y, X, w=W)

importance_partial_r2 = partial_r2(model, y, X, w=W)
```

```python
from scipy.stats import norm

X = df_analysis[svi_measures]
num_cols = X.select_dtypes(include=[np.number]).columns

fitness_dict = {}

for col in num_cols:
    mu = df_train_states_svi[col].mean()
    sigma = df_train_states_svi[col].std(ddof=1)

    # 1) z-score fitness (how many stds away from B's mean)
    z = (X[col] - mu) / sigma

    # 2) density fitness (higher = more typical in B)
    pdf = norm.pdf(X[col], loc=mu, scale=sigma)

    fitness_dict[col + "_zfit"] = z
    fitness_dict[col + "_pdffit"] = pdf

X_fitness = pd.DataFrame(fitness_dict, index=X.index)
```

```python
X = X_fitness[ [x + "_pdffit" for x in svi_measures]]
y = df_analysis['CPC']
# X = gdf_analysis_per[svi_measures]
# y = gdf_analysis_per['CPC']

# Add constant (intercept)
X = sm.add_constant(X)

# Fit model
model = sm.OLS(y, X).fit()

# Output regression summary
print(model.summary())
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from scipy.stats import multivariate_normal
import numpy as np

# X: reference dataset, x_new: new data point (1D array)
scaler = StandardScaler().fit(df_train_states_svi[svi_measures])
X_scaled = scaler.transform(df_train_states_svi[svi_measures])
x_new_scaled = scaler.transform(df_analysis[svi_measures])

# Now safe to compute distances / density
mu = X_scaled.mean(axis=0)
cov = np.cov(X_scaled, rowvar=False)
pdf_value = multivariate_normal.pdf(x_new_scaled, mean=mu, cov=cov)
```
