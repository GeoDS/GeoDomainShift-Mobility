```python
import pandas as pd
import os
import random

import numpy as np
from sklearn.preprocessing import StandardScaler
import geopandas as gpd

from spreg import GM_Combo, gets_sdm, GM_Lag
import libpysal
from spreg import OLS
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def load_state_areas(if_shuffle=True, state_codes:list = None):
    # areas = os.listdir("data")
    areas = [area for area in os.listdir("../../../data") if area.startswith(tuple(state_codes))]
    if if_shuffle:
        random.shuffle(areas)
    return areas
```

```python
train_states = ['04']
```

```python
train_areas = load_state_areas(if_shuffle=False, state_codes = train_states)
```

```python
print(train_areas)
```

```python
def load_area_features(areas):
    demos_list = []
    pois_list = []
    dis_list = []
    for area in areas:
        if area == ".DS_Store":
            continue
        demos = np.load(f"../../../data/{area}/demos.npy")
        pois = np.load(f"../../../data/{area}/pois.npy")
        # dis = np.load(f"../../../data/{area}/dis.npy")
        
        demos_list.append(demos)
        pois_list.append(pois)
        # dis_list.append(dis)

    demos = np.concatenate(demos_list, axis=0)
    pois = np.concatenate(pois_list, axis=0)
    # dis = np.concatenate(dis_list)

    return demos, pois
```

```python
# feat_scaler = StandardScaler()
# demos = feat_scaler.fit_transform(demos)
# pois = feat_scaler.fit_transform(pois)
# dis = feat_scaler.fit_transform(dis)
src_demos, src_pois = load_area_features(train_areas)
```

```python
df_us_states = pd.read_csv("../data/us_states_fips.csv", dtype={"FIPS": str})
test_states = list(set(df_us_states['FIPS']) - set(train_states))
test_areas = load_state_areas(if_shuffle=False, state_codes = test_states)
```

```python
def src_tar_diff(src_demos, src_pois, test_areas):
    rows = []
    for area in test_areas:
        tar_demos, tar_pois = load_area_features([area])

        demo_mean_diff = np.abs(src_demos.mean(0) - tar_demos.mean(0))
        poi_mean_diff = np.abs(src_pois.mean(0) - tar_pois.mean(0))

        demo_smd = (src_demos.mean(0) - tar_demos.mean(0))/(src_demos.std(0) + 1e-6)
        poi_smd = (src_pois.mean(0) - tar_pois.mean(0))/(src_pois.std(0) + 1e-6)

        row = {'GEOID': area}
        keys = ['demo_mean_diff_{}'.format(i) for i in range(len(demo_mean_diff))]
        d = dict(zip(keys, demo_mean_diff))
        row.update(d)

        keys = ['poi_mean_diff_{}'.format(i) for i in range(len(poi_mean_diff))]
        d = dict(zip(keys, poi_mean_diff))
        row.update(d)

        keys = ['demo_smd_{}'.format(i) for i in range(len(demo_smd))]
        d = dict(zip(keys, demo_smd))
        row.update(d)

        keys = ['poi_smd_{}'.format(i) for i in range(len(poi_smd))]
        d = dict(zip(keys, poi_smd))
        row.update(d)

        rows.append(row)

    df_src_tar_diff = pd.DataFrame(rows)
    return df_src_tar_diff
```

```python
# demo_mean_diffs = []
# poi_mean_diffs = []
# dis_mean_diffs = []

# demo_smds = []
# poi_smds = []
# dis_smds = []
rows = []
for area in test_areas:
    tar_demos, tar_pois = load_area_features([area])

    demo_mean_diff = np.abs(src_demos.mean(0) - tar_demos.mean(0))
    poi_mean_diff = np.abs(src_pois.mean(0) - tar_pois.mean(0))
    # dis_mean_diff = np.abs(src_dis.mean(0) - tar_dis.mean(0))

    demo_smd = (src_demos.mean(0) - tar_demos.mean(0))/(src_demos.std(0) + 1e-6)
    poi_smd = (src_pois.mean(0) - tar_pois.mean(0))/(src_pois.std(0) + 1e-6)
    # dis_smd = (src_dis.mean(0) - tar_dis.mean(0))/src_dis.std(0)

    row = {'GEOID': area}
    keys = ['demo_mean_diff_{}'.format(i) for i in range(len(demo_mean_diff))]
    d = dict(zip(keys, demo_mean_diff))
    row.update(d)

    keys = ['poi_mean_diff_{}'.format(i) for i in range(len(poi_mean_diff))]
    d = dict(zip(keys, poi_mean_diff))
    row.update(d)
    # demo_mean_diffs.append(emo_mean_diff)
    # poi_mean_diffs.append(poi_mean_diff)
    # dis_mean_diffs.append(dis_mean_diff)

    keys = ['demo_smd_{}'.format(i) for i in range(len(demo_smd))]
    d = dict(zip(keys, demo_smd))
    row.update(d)
    # demo_smds.append(demo_smd)

    keys = ['poi_smd_{}'.format(i) for i in range(len(poi_smd))]
    d = dict(zip(keys, poi_smd))
    row.update(d)
    # poi_smds.append(poi_smd)
    # dis_mean_smds.append(dis_smd)

    rows.append(row)
```

### Add county-level performance data

```python
# us_county_geo_file = '../data/tl_2024_us_county/tl_2024_us_county.shp'
us_county_geo_file = '../data/tl_2018_us_county/tl_2018_us_county.shp'
```

```python
gpd_us_county =  gpd.read_file(us_county_geo_file)
```

```python
gpd_us_county  = gpd_us_county [~gpd_us_county['STATEFP'].isin(['02','15','60','66','69','72','78'])]
```

```python
eval_results = pd.read_csv('tables/source_{}_county_level.csv'.format('+'.join(train_states)), dtype={"GEOID": str})
```

```python
eval_results = eval_results[~eval_results['GEOID'].str.startswith(('02','15','60','66','69','72','78'))]
```

```python
gpd_eval_results = pd.merge(eval_results, gpd_us_county[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry']], on="GEOID", how="left")
```

```python
gpd_eval_results.shape
```

```python
gpd_eval_results = gpd.GeoDataFrame(gpd_eval_results, geometry="geometry")
```

```python
df_analysis = pd.merge(gpd_eval_results[gpd_eval_results['CPC'] != -1], df_src_tar_diff, left_on="GEOID", right_on="GEOID", how="left")
```

```python
df_analysis.shape
```

```python
predictors = ['demo_mean_diff_{}'.format(i) for i in range(src_demos.shape[1])] + ['poi_mean_diff_{}'.format(i) for i in range(src_pois.shape[1])]
# predictors = ['poi_mean_diff_{}'.format(i) for i in range(src_pois.shape[1])]
# predictors = ['demo_smd_{}'.format(i) for i in range(src_demos.shape[1])]
# predictors = ['poi_smd_{}'.format(i) for i in range(src_pois.shape[1])]
```

```python
metrics = ['CPC', 'RMSE', 'NRMSE', 'JSD_inflow', 'JSD_outflow', 'JSD_ODflow']
```

```python
df_analysis_per = df_analysis[metrics + predictors+ ['geometry']]
```

```python
feat_scaler = StandardScaler()

gdf_analysis_per = gpd.GeoDataFrame(df_analysis_per, geometry="geometry")
gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
gdf_analysis_per = gdf_analysis_per.to_crs(epsg=5070)
gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
coords = np.array([gdf_analysis_per.centroid.x, gdf_analysis_per.centroid.y]).T
    
W = libpysal.weights.KNN.from_array(coords, k=8)
W.transform = "r"
    
X = gdf_analysis_per[predictors]
X = feat_scaler.fit_transform(X)
# rows_with_null = X[X.isnull().any(axis=1)]
# print(rows_with_null)
y = gdf_analysis_per['RMSE'].apply(np.log1p).values.reshape((-1,1))
# y = gdf_analysis_per['CPC'].values.reshape((-1,1))
model = gets_sdm(y, X, w=W)[1]

ols = OLS(y, X)
```

```python
from esda.moran import Moran

mi_lag = Moran(model.u, W)
print("Spatial Model Moran I:", mi_lag.I, "p-value:", mi_lag.p_norm)

mi_ols = Moran(ols.u, W)
print("OLS Moran I:", mi_ols .I, "p-value:", mi_ols.p_norm)
```

```python
rho = model.betas[-1]
coef = model.output.iloc[:-1, :].copy()
coef['total_impact'] = coef.loc[:, 'coefficients']/ rho
coef['source'] = ['+'.join(train_states)] * coef.shape[0]
```

### Store spatial autoregressive model results from multiple train datasets

```python
from spreg import GM_Combo, gets_sdm, GM_Lag, GM_Combo_Het
import libpysal
from spreg import OLS

source_states = [['01'], ['04'], ['05'], ['06'], ['08'], ['09'], ['10'], ['12'], ['13'], ['17'], ['18'], ['19'], ['20'], ['21'], ['22'], ['36']]
# source_states = [['19']]
# svi_file = '../data/SVI_2018_US_county.csv'
# df_svi = pd.read_csv(svi_file, dtype={"ST": str, "FIPS": str, "E_TOTPOP": int})
feature_type = 'demo'
```

```python
coef_dfs = []

states = []
lag_pr2s = []
lag_pr2_es = []
ols_r2s = []
lag_morans = []
lag_moran_pvals = []
ols_morans = []
ols_moran_pvals = []

for train_states in source_states:
    print('+'.join(train_states))

    states.append('+'.join(train_states))
    
    train_areas = load_state_areas(if_shuffle=False, state_codes = train_states)
    src_demos, src_pois = load_area_features(train_areas)

    df_us_states = pd.read_csv("../data/us_states_fips.csv", dtype={"FIPS": str})
    test_states = list(set(df_us_states['FIPS']) - set(train_states))
    test_areas = load_state_areas(if_shuffle=False, state_codes = test_states)

    df_src_tar_diff = src_tar_diff(src_demos, src_pois, test_areas)

    us_county_geo_file = '../data/tl_2018_us_county/tl_2018_us_county.shp'
    gpd_us_county =  gpd.read_file(us_county_geo_file)
    gpd_us_county  = gpd_us_county [~gpd_us_county['STATEFP'].isin(['02','15','60','66','69','72','78'])]
    
    eval_results = pd.read_csv('tables/source_{}_county_level.csv'.format('+'.join(train_states)), dtype={"GEOID": str})
    eval_results = eval_results[~eval_results['GEOID'].str.startswith(('02','15','60','66','69','72','78'))]
    avg_eval = eval_results.drop(columns = ['GEOID', 'num_regions'])

    gpd_eval_results = pd.merge(eval_results, gpd_us_county[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry']], on="GEOID", how="left")
    metrics = ['CPC', 'RMSE', 'NRMSE', 'JSD_inflow', 'JSD_outflow', 'JSD_ODflow']
    gpd_eval_results_vis = gpd_eval_results[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry'] + metrics]
    # gpd_eval_results_vis = pd.concat([gpd_eval_results_vis, pd.DataFrame(state_rows)], ignore_index=True)
    # gpd_eval_results_vis[metrics] = gpd_eval_results_vis[metrics].fillna(-1)
    gpd_eval_results_vis = gpd.GeoDataFrame(gpd_eval_results_vis, geometry="geometry")

    df_analysis = pd.merge(gpd_eval_results_vis[gpd_eval_results_vis['CPC'] != -1], df_src_tar_diff, left_on="GEOID", right_on="GEOID", how="left")
    predictors = ['{}_mean_diff_{}'.format(feature_type, i) for i in range(src_demos.shape[1])]
    # predictors =  ['poi_mean_diff_{}'.format(i) for i in range(src_pois.shape[1])]
    df_analysis_per = df_analysis[metrics + predictors + ['geometry']]

    df_analysis_per[predictors] = feat_scaler.fit_transform(df_analysis_per[predictors] )
    
    gdf_analysis_per = gpd.GeoDataFrame(df_analysis_per, geometry="geometry")
    gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
    gdf_analysis_per = gdf_analysis_per.to_crs(epsg=5070)
    gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
    coords = np.array([gdf_analysis_per.centroid.x, gdf_analysis_per.centroid.y]).T
    
    W = libpysal.weights.KNN.from_array(coords, k=8)
    W.transform = "r"
    
    X = gdf_analysis_per[predictors]

    # rows_with_null = X[X.isnull().any(axis=1)]
    # print(rows_with_null)
    y = gdf_analysis_per['RMSE'].apply(np.log1p).values.reshape((-1,1))
    # y = gdf_analysis_per['CPC'].values.reshape((-1,1))

    # model = gets_sdm(y, X, w=W)[1]
    model = GM_Combo_Het(y, X, w=W)  

    ols = OLS(y, X)

    rho = model.betas[-1]
    coef = model.output.iloc[:-1, :].copy()
    coef['total_impact'] = coef.loc[:, 'coefficients']/ rho
    coef['source'] = ['+'.join(train_states)] * coef.shape[0]
    coef_dfs.append(coef)

    lag_pr2s.append(model.pr2)
    lag_pr2_es.append(model.pr2_e)
    ols_r2s.append(ols.r2)

    mi_lag = Moran(model.u, W)
    lag_morans.append(mi_lag.I)
    lag_moran_pvals.append(mi_lag.p_norm)

    mi_ols = Moran(ols.u, W)
    ols_morans.append(mi_ols.I)
    ols_moran_pvals.append(mi_ols.p_norm)
```

```python
df_coefs = pd.concat(coef_dfs, ignore_index=True)
```

```python
df_coefs[['coefficients','std_err','zt_stat','prob']] = df_coefs[['coefficients','std_err','zt_stat','prob']].apply(pd.to_numeric, errors='coerce')
```

```python
df_coefs[['coefficients','std_err','zt_stat','prob']] = df_coefs[['coefficients','std_err','zt_stat','prob']].round(4)
```

```python
df_coefs.to_csv('{}_coefficients.csv'.format(feature_type))
```

```python
df_coefs = df_coefs[df_coefs['var_names'] != 'CONSTANT']
df_coefs = df_coefs[df_coefs['prob'] < 0.05]
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 40))

ax = sns.pointplot(
    data=df_coefs,
    x='total_impact',
    y='var_names',
    hue='source',
    dodge=0.6,
    capsize=0.1,
    # ci=None, # Important: disable built-in CI
    # palette=palette_colors,
    # markers=['o', 's', '^'], # Custom markers for each model
    linestyles=[''] * df_coefs.shape[0]# Ensure no lines are connecting the markers
)

# This is the vertical line at 0
plt.axvline(x=0, color='red', linestyle='--', linewidth=1.5, alpha=0.7)

# Ensure the horizontal grid lines are visible and align with the variables
ax.yaxis.grid(True, linestyle='-', alpha=0.6)
ax.xaxis.grid(True, linestyle='--')

# 1. Get the current major tick locations (y-positions of the variables)
current_ticks = ax.get_yticks()

# 2. Calculate the intermediate positions for the grid lines.
# We want the lines to sit halfway between the variable groups.
# We skip the first and last position for the grid, as they are typically plot borders.
grid_ticks = current_ticks[:-1] + np.diff(current_ticks) / 2

# 3. Apply the custom grid lines to the axes
ax.yaxis.grid(False) # Turn off the default y-grid first
ax.set_yticks(grid_ticks, minor=True) # Set the new positions as minor ticks
ax.grid(which='minor', axis='y', linestyle='-', alpha=0.6, color='gray') # Draw the grid using minor ticks

# 4. Restore the original major ticks (where the variable labels sit)
ax.set_yticks(current_ticks)
ax.set_yticklabels(df_coefs['var_names'].unique().tolist())

# plt.tight_layout()
plt.savefig('{}_total_impact_forest_plot.pdf'.format(feature_type), bbox_inches='tight')
plt.show()
```
