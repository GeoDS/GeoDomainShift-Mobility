```python
import numpy as np
from scipy.sparse.linalg import inv
from scipy.sparse import identity

def total_impacts(model, w=None):
    n_x = len(model.name_x)
    betas = model.betas.flatten()
    # [1:1+n_x]  # drop intercept

    impacts = []

    # Case 1: models with rho (SAR / SARAR / SDM etc.)
    if hasattr(model, "rho"):
        from scipy.sparse import identity
        from scipy.sparse.linalg import inv

        n = w.n
        W = w.sparse
        rho = float(model.rho)

        I = identity(n)
        S = inv(I - rho * W)

        trS = S.diagonal().sum()
        sumS = S.sum()

        for i, b in enumerate(betas):
            direct = (trS / n) * b
            indirect = ((sumS - trS) / n) * b
            total = direct + indirect

            # impacts[model.name_x[i]] = {
            #     "direct": direct,
            #     "indirect": indirect,
            #     "total": total,
            # }
            impacts.append(total)

    # Case 2: error-only models (GMM_Error, GM_Error, SEM)
    else:
        for i, b in enumerate(betas):
            impacts.append(b)
            # impacts[model.name_x[i]] = {
            #     "direct": b,
            #     "indirect": 0.0,
            #     "total": b,
            # }

    return impacts
```

```python
from scipy.stats import skew

metrics = ['CPC', 'RMSE', 'NRMSE', 'JSD_inflow', 'JSD_outflow', 'JSD_ODflow']
predictors = ['mi_shift', 'moran_shift']

coef_dfs = []

states = []
lag_pr2s = []
# lag_pr2_es = []
ols_r2s = []
lag_morans = []
lag_moran_pvals = []
ols_morans = []
ols_moran_pvals = []

for train_states in source_states:
    states.append('+'.join(train_states))

    test_states = list(set(df_us_states['FIPS']) - set(train_states))

    df_domain_shift = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_domain_shift = df_domain_shift.rename(columns = {'moran_shift': 'moran_dis'})

    df_domain_shift_v2 = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state_moran/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_src_tar_shift = df_domain_shift.copy().merge(df_domain_shift_v2, on=['src_state', 'tar_geoid'], how='left' )
    df_src_tar_shift = df_src_tar_shift.set_index('tar_geoid')[['mi_shift',	'moran_shift']]

    eval_results = pd.read_csv('models/DGM_raw/evaluations/tables/test/source_{}_county_level.csv'.format('+'.join(train_states)), dtype={"GEOID": str})
    eval_results = eval_results[eval_results['GEOID'].str.startswith(tuple(test_states))]
    avg_eval = eval_results.drop(columns = ['GEOID', 'num_regions'])

    gpd_eval_results = pd.merge(eval_results, gpd_us_county[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry']], on="GEOID", how="left")

    gpd_eval_results_vis = gpd_eval_results[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry'] + metrics]
    gpd_eval_results_vis = gpd.GeoDataFrame(gpd_eval_results_vis, geometry="geometry")

    df_analysis = pd.merge(gpd_eval_results_vis[gpd_eval_results_vis['CPC'] != -1], df_src_tar_shift, left_on="GEOID", right_index=True, how="left")
    # df_analysis = df_analysis.dropna(subset=["CPC"] + predictors + ["geometry"]).copy()

    feat_scaler = StandardScaler()

    df_analysis_per = df_analysis.dropna(subset= metrics + predictors + ['geometry']).copy()
    df_analysis_per[predictors] = feat_scaler.fit_transform(df_analysis_per[predictors])

    gdf_analysis_per = gpd.GeoDataFrame(df_analysis_per, geometry="geometry")
    gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
    gdf_analysis_per = gdf_analysis_per.to_crs(epsg=5070)
    gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
    coords = np.array([gdf_analysis_per.centroid.x, gdf_analysis_per.centroid.y]).T

    W = libpysal.weights.KNN.from_array(coords, k=8)
    W.transform = "r"

    # print("islands:", len(W.islands))

    X = gdf_analysis_per[predictors]

    # y = gdf_analysis_per['RMSE'].apply(np.log1p).values.reshape((-1,1))
    y = gdf_analysis_per['CPC'].values.reshape((-1,1))

    try:
      model = gets_sdm(y, X, w=W)[1]
      model_type = "SDM"

    except Exception as e:
      print(f"[SDM failed] source={'+'.join(train_states)} | {e}")

    try:
        # Fallback: SAR via GMM/2SLS
        model = GM_Lag(
            y, X, w=W,
            # optional but recommended:
            # name_y="CPC", name_x=["const"] + predictors
        )
        model_type = "GM_Lag"

    except Exception as e2:
        print(f"[GM_Lag failed] source={'+'.join(train_states)} | {e2}")
        continue   # skip this iteration entirely



    # model = gets_sdm(y, X, w=W)[0]
    # ols = OLS(y, X)
    print(model.summary)
    moran_metric = Moran(model.u, W)
    print("Moran'I is: {}, sig: {}".format(moran_metric.I, moran_metric.p_norm))

```

```python
from spreg import ML_Lag

metrics = ['CPC', 'RMSE', 'NRMSE', 'JSD_inflow', 'JSD_outflow', 'JSD_ODflow']
predictors = ['mi_shift', 'moran_shift']
# , 'moran_dis'
# predictors = ['mi_shift','tar_moran', 'nn_moran']

coef_dfs = []

states = []
lag_pr2s = []
# lag_pr2_es = []
ols_r2s = []
lag_morans = []
lag_moran_pvals = []
ols_morans = []
ols_moran_pvals = []

# def nn_signed_diff(src_arr: np.ndarray, tar: float) -> float:
#     d = tar - src_arr                  # signed diffs to each source value
#     return d[np.argmin(np.abs(d))]     # nearest by abs, keep sign

for train_states in source_states:
    states.append('+'.join(train_states))

    test_states = list(set(df_us_states['FIPS']) - set(train_states))

    df_domain_shift = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_domain_shift = df_domain_shift.rename(columns = {'moran_shift': 'moran_dis'})

    df_domain_shift_v2 = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state_moran/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_domain_shift_v2["src_moran"] = df_domain_shift_v2["src_moran"].apply(
        lambda s: np.fromstring(str(s).strip("[]"), sep=" ")
    )

    df_domain_shift_v2["tar_moran"] = df_domain_shift_v2["tar_moran"].apply(
        lambda s: np.fromstring(str(s).strip("[]"), sep=" ").item()
    )

    # Calculate mean_moran element-wise for each row
    # df_domain_shift_v2["mean_moran"] = df_domain_shift_v2.apply(
    #     lambda row: np.mean(np.abs(row['tar_moran'] - row['src_moran']))
    #                 if isinstance(row['src_moran'], np.ndarray) and row['src_moran'].size > 0
    #                 else np.nan,
    #     axis=1
    # )
    # df_domain_shift_v2["nn_moran"] = df_domain_shift_v2.apply(
    #     lambda r: nn_signed_diff(r["src_moran"], r["tar_moran"]),
    #     axis=1
    # )
    # df_domain_shift_v2["src_moran_mean"] = df_domain_shift_v2["src_moran"].apply(np.mean)
    # df_domain_shift_v2["src_moran_std"]  = df_domain_shift_v2["src_moran"].apply(np.std)
    df_domain_shift_v2["moran_mean"]  = df_domain_shift_v2["tar_moran"] - df_domain_shift_v2["src_moran"].apply(np.mean)

    df_src_tar_shift = df_domain_shift.copy().merge(df_domain_shift_v2, on=['src_state', 'tar_geoid'], how='left')

    df_src_tar_shift = df_src_tar_shift.set_index('tar_geoid')[predictors]

    eval_results = pd.read_csv('models/DGM_raw/evaluations/tables/test/source_{}_county_level.csv'.format('+'.join(train_states)), dtype={"GEOID": str})
    eval_results = eval_results[eval_results['GEOID'].str.startswith(tuple(test_states))]
    avg_eval = eval_results.drop(columns = ['GEOID', 'num_regions'])

    gpd_eval_results = pd.merge(eval_results, gpd_us_county[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry']], on="GEOID", how="left")

    gpd_eval_results_vis = gpd_eval_results[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry'] + metrics]
    gpd_eval_results_vis = gpd.GeoDataFrame(gpd_eval_results_vis, geometry="geometry")

    df_analysis = pd.merge(gpd_eval_results_vis[gpd_eval_results_vis['CPC'] != -1], df_src_tar_shift, left_on="GEOID", right_index=True, how="left")

    feat_scaler = StandardScaler()

    df_analysis_per = df_analysis.dropna(subset= metrics + predictors + ['geometry']).copy()

    # df_analysis_per['moran_shift'] = df_analysis_per['moran_shift']
    df_analysis_per[predictors] = feat_scaler.fit_transform(df_analysis_per[predictors])

    gdf_analysis_per = gpd.GeoDataFrame(df_analysis_per, geometry="geometry")
    gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
    gdf_analysis_per = gdf_analysis_per.to_crs(epsg=5070)
    gdf_analysis_per = gdf_analysis_per[~gdf_analysis_per.geometry.isna()]
    coords = np.array([gdf_analysis_per.centroid.x, gdf_analysis_per.centroid.y]).T

    W = libpysal.weights.KNN.from_array(coords, k=8)
    W.transform = "r"

    X = gdf_analysis_per[predictors]

    # y = gdf_analysis_per['RMSE'].apply(np.log1p).values.reshape((-1,1))
    # model = gets_sdm(y, X, w=W)[1]

    y = gdf_analysis_per['CPC'].values.reshape((-1,1))
    model = ML_Lag(
            y, X, w=W, slx_lags = 1, spat_diag=True, vif=True,
        )

    ols = OLS(y, X)

    coef = model.output.copy()
    # rho = model.betas[-1]
    # coef['total_impact'] = coef.loc[:, 'coefficients']/ rho
    coef['total_impact'] = total_impacts(model, W)
    coef['source'] = ['+'.join(train_states)] * coef.shape[0]
    coef_dfs.append(coef)

    lag_pr2s.append(model.pr2)
    # lag_pr2s.append(model.r2)
    # lag_pr2_es.append(model.pr2_e)
    ols_r2s.append(ols.r2)

    mi_lag = Moran(model.u, W)
    lag_morans.append(mi_lag.I)
    lag_moran_pvals.append(mi_lag.p_norm)

    mi_ols = Moran(ols.u, W)
    ols_morans.append(mi_ols.I)
    ols_moran_pvals.append(mi_ols.p_norm)
```

```python
import numpy as np
import ast # Required for literal_eval

src_tar_shift_list = []
eval_results_list = []

def nn_signed_diff(src_arr: np.ndarray, tar: float) -> float:
    d = tar - src_arr                  # signed diffs to each source value
    return d[np.argmin(np.abs(d))]     # nearest by abs, keep sign

for train_states in source_states:
    # states.append('+'.join(train_states))

    test_states = list(set(df_us_states['FIPS']) - set(train_states))

    df_domain_shift = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_domain_shift = df_domain_shift.rename(columns = {'moran_shift': 'moran_dis'})

    df_domain_shift_v2 = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state_moran/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_domain_shift_v2["src_moran"] = df_domain_shift_v2["src_moran"].apply(
        lambda s: np.fromstring(str(s).strip("[]"), sep=" ")
    )

    df_domain_shift_v2["tar_moran"] = df_domain_shift_v2["tar_moran"].apply(
        lambda s: np.fromstring(str(s).strip("[]"), sep=" ").item()
    )

    # df_domain_shift_v2["nn_moran"] = df_domain_shift_v2.apply(
    #     lambda r: nn_signed_diff(r["src_moran"], r["tar_moran"]),
    #     axis=1
    # )
    df_domain_shift_v2["src_moran_std"]  = df_domain_shift_v2["src_moran"].apply(np.std)
    # df_domain_shift_v2["src_moran_cv"]  = df_domain_shift_v2["src_moran"].apply(np.std) / df_domain_shift_v2["src_moran"].apply(np.mean)
    # df_domain_shift_v2["src_moran"] = df_domain_shift_v2["src_moran"].apply(np.mean)
    df_domain_shift_v2["moran_mean"]  = df_domain_shift_v2["tar_moran"] - df_domain_shift_v2["src_moran"].apply(np.mean)

    df_src_tar_shift = df_domain_shift.copy().merge(df_domain_shift_v2, on=['src_state', 'tar_geoid'], how='left')
    # Set 'tar_geoid' as the index before appending to the list
    # df_src_tar_shift = df_src_tar_shift.set_index('tar_geoid')

    eval_results = pd.read_csv('models/DGM_raw/evaluations/tables/test/source_{}_county_level.csv'.format('+'.join(train_states)), dtype={"GEOID": str})
    eval_results = eval_results[eval_results['GEOID'].str.startswith(tuple(test_states))]
    eval_results['src_state'] = ['+'.join(train_states)] * eval_results.shape[0]
    # avg_eval = eval_results.drop(columns = ['GEOID', 'num_regions'])

    src_tar_shift_list.append(df_src_tar_shift)
    eval_results_list.append(eval_results)

df_src_tar_shifts = pd.concat(src_tar_shift_list)
df_eval_results = pd.concat(eval_results_list, ignore_index=True)
```

#### **2.1 Moran's I of model performance**

```python
metrics = ['CPC', 'RMSE', 'NRMSE', 'JSD_inflow', 'JSD_outflow', 'JSD_ODflow']

src_states = []
cpc_moran_list = []
rmse_moran_list = []
cpc_moran_p = []
rmse_moran_p = []

for train_states in source_states:
    src_states.append('+'.join(train_states))

    test_states = list(set(df_us_states['FIPS']) - set(train_states))

    df_domain_shift = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_domain_shift = df_domain_shift.rename(columns = {'moran_shift': 'moran_dis'})

    df_domain_shift_v2 = pd.read_csv('models/DGM_raw/evaluations/domain_shift/by_source_state_moran_shift/{}_domain_shift.csv'.format('+'.join(train_states)),
                                  dtype={"src_state": str, "tar_geoid": str})

    df_src_tar_shift = df_domain_shift.copy().merge(df_domain_shift_v2, on=['src_state', 'tar_geoid'], how='left')

    df_src_tar_shift = df_src_tar_shift.set_index('tar_geoid')

    eval_results = pd.read_csv('models/DGM_raw/evaluations/tables/test/source_{}_county_level.csv'.format('+'.join(train_states)), dtype={"GEOID": str})
    eval_results = eval_results[eval_results['GEOID'].str.startswith(tuple(test_states))]
    avg_eval = eval_results.drop(columns = ['GEOID', 'num_regions'])

    gpd_eval_results = pd.merge(eval_results, gpd_us_county[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry']], on="GEOID", how="left")

    gpd_eval_results_vis = gpd_eval_results[['STATEFP', 'COUNTYFP', 'GEOID', 'NAME', 'geometry'] + metrics]
    gpd_eval_results_vis = gpd.GeoDataFrame(gpd_eval_results_vis, geometry="geometry")
    gpd_eval_results_vis = gpd_eval_results_vis[~gpd_eval_results_vis.geometry.isna()]
    gpd_eval_results_vis = gpd_eval_results_vis.to_crs(epsg=5070)
    coords = np.array([gpd_eval_results_vis.centroid.x, gpd_eval_results_vis.centroid.y]).T

    W = libpysal.weights.KNN.from_array(coords, k=8)
    W.transform = "r"

    cpc_moran = Moran(gpd_eval_results_vis['CPC'], W)
    cpc_moran_list.append(cpc_moran.I)
    cpc_moran_p.append(cpc_moran.p_norm)

    rmse_moran = Moran(gpd_eval_results_vis['RMSE'], W)
    rmse_moran_list.append(rmse_moran.I)
    rmse_moran_p.append(rmse_moran.p_norm)

df_src_moran = pd.DataFrame({'src_state': src_states, 'cpc_moran': cpc_moran_list, 'cpc_moran_p': cpc_moran_p,
                             'rmse_moran': rmse_moran_list, 'rmse_moran_p': rmse_moran_p})
```

```python
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
from pygam import LinearGAM, s
import statsmodels.api as sm

X = df_analysis_agg_src['src_moran_mean'].values.reshape(-1, 1)
y = df_analysis_agg_src['CPC_mean']

kf = KFold(n_splits=5, shuffle=True, random_state=0)

rmse_lin, rmse_quad, rmse_gam = [], [], []

for tr, te in kf.split(X):
    Xtr, Xte = X[tr], X[te]
    ytr, yte = y[tr], y[te]

    # Linear
    Xtr_lin = sm.add_constant(Xtr)
    Xte_lin = sm.add_constant(Xte)
    lin = sm.OLS(ytr, Xtr_lin).fit()
    pred_lin = lin.predict(Xte_lin)
    rmse_lin.append(np.sqrt(mean_squared_error(yte, pred_lin)))

    # Quadratic
    Xtr_q = sm.add_constant(np.c_[Xtr, Xtr**2])
    Xte_q = sm.add_constant(np.c_[Xte, Xte**2])
    quad = sm.OLS(ytr, Xtr_q).fit()
    pred_quad = quad.predict(Xte_q)
    rmse_quad.append(np.sqrt(mean_squared_error(yte, pred_quad)))

    # GAM (keep low splines to discourage wiggles)
    gam = LinearGAM(s(0, n_splines=5)).gridsearch(Xtr, ytr)
    pred_gam = gam.predict(Xte)
    rmse_gam.append(np.sqrt(mean_squared_error(yte, pred_gam)))

print("CV RMSE (mean):")
print("  Linear   :", np.mean(rmse_lin))
print("  Quadratic:", np.mean(rmse_quad))
print("  GAM      :", np.mean(rmse_gam))
```

```python
import numpy as np
import pandas as pd
import patsy
import matplotlib.pyplot as plt
import seaborn as sns

mpl.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "legend.fontsize": 12,
    "axes.labelsize": 14,
    "pdf.fonttype": 42,
})

def joint_partial_heatmap(
    df,
    result,
    mi_col="mi_shift",
    moran_col="moran_shift",
    state_col="src_state",
    rhs="C(src_state) + mi_shift + moran_shift",  # MUST match your MixedLM fixed-effects RHS
    n_grid=60,
    use_state="mode",   # "mode" or specify a value like "CA" / 04 etc.
    q_low=0.05,
    q_high=0.95,
    cmap="viridis"
):
    # --- 1) Choose ranges (avoid extreme tails) ---
    mi_grid = np.linspace(df[mi_col].quantile(q_low), df[mi_col].quantile(q_high), n_grid)
    moran_grid = np.linspace(df[moran_col].quantile(q_low), df[moran_col].quantile(q_high), n_grid)
    MI, MORAN = np.meshgrid(mi_grid, moran_grid)

    # --- 2) Pick a reference training state for the fixed effects ---
    if use_state == "mode":
        ref_state = df[state_col].mode()[0]
    else:
        ref_state = use_state

    # Force category levels to match training data (important for C(src_state))
    src_cats = pd.Categorical(df[state_col]).categories

    pred_df = pd.DataFrame({
        mi_col: MI.ravel(),
        moran_col: MORAN.ravel(),
        state_col: pd.Categorical([ref_state] * MI.size, categories=src_cats)
    })

    # --- 3) Build X using the SAME columns as the fitted model ---
    X_pred = patsy.dmatrix(rhs, pred_df, return_type="dataframe")

    beta = result.fe_params  # pandas Series
    X_pred = X_pred.reindex(columns=beta.index, fill_value=0.0)  # align exactly

    Z = (X_pred.to_numpy() @ beta.to_numpy()).reshape(MI.shape)

    # --- 4) Plot (heatmap style) ---
    plt.figure(figsize=(8, 6))
    ax = sns.heatmap(
        Z,
        cmap=cmap,
        cbar_kws={"label": "Predicted CPC (fixed effects only)"}
    )

    # Make axes readable (don’t label every grid point)
    xticks = np.linspace(0, n_grid - 1, 6).astype(int)
    yticks = np.linspace(0, n_grid - 1, 6).astype(int)

    ax.set_xticks(xticks)
    ax.set_yticks(yticks)

    ax.set_xticklabels([f"{mi_grid[i]:.2f}" for i in xticks], rotation=0)
    ax.set_yticklabels([f"{moran_grid[i]:.2f}" for i in yticks], rotation=0)

    ax.set_xlabel("MI Shift")
    ax.set_ylabel("Moran Shift")
    # ax.set_title("Joint partial effect of MI Shift and Moran Shift (MixedLM)")

    ax.invert_yaxis()
    plt.tight_layout()
    # plt.savefig('models/DGM_raw/evaluations/figures/Joint_partial_effects_CPC.pdf', bbox_inches='tight')
    plt.show()

    return MI, MORAN, Z
```

```python
import numpy as np
import pandas as pd
import patsy
import matplotlib.pyplot as plt
import seaborn as sns

mpl.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "legend.fontsize": 12,
    "axes.labelsize": 14,
    "pdf.fonttype": 42,
})

def joint_partial_heatmap(
    df,
    result,
    mi_col="mi_shift",
    moran_col="moran_shift",
    state_col="src_state",
    rhs="C(src_state) + mi_shift + moran_shift",  # MUST match your MixedLM fixed-effects RHS
    n_grid=60,
    use_state="mode",   # "mode" or specify a value like "CA" / 04 etc.
    q_low=0.05,
    q_high=0.95,
    cmap="viridis_r"
):
    # --- 1) Choose ranges (avoid extreme tails) ---
    mi_grid = np.linspace(df[mi_col].quantile(q_low), df[mi_col].quantile(q_high), n_grid)
    moran_grid = np.linspace(df[moran_col].quantile(q_low), df[moran_col].quantile(q_high), n_grid)
    MI, MORAN = np.meshgrid(mi_grid, moran_grid)

    # --- 2) Pick a reference training state for the fixed effects ---
    if use_state == "mode":
        ref_state = df[state_col].mode()[0]
    else:
        ref_state = use_state

    # Force category levels to match training data (important for C(src_state))
    src_cats = pd.Categorical(df[state_col]).categories

    pred_df = pd.DataFrame({
        mi_col: MI.ravel(),
        moran_col: MORAN.ravel(),
        state_col: pd.Categorical([ref_state] * MI.size, categories=src_cats)
    })

    # --- 3) Build X using the SAME columns as the fitted model ---
    X_pred = patsy.dmatrix(rhs, pred_df, return_type="dataframe")

    beta = result.fe_params  # pandas Series
    X_pred = X_pred.reindex(columns=beta.index, fill_value=0.0)  # align exactly

    Z = (X_pred.to_numpy() @ beta.to_numpy()).reshape(MI.shape)

    # --- 4) Plot (heatmap style) ---
    plt.figure(figsize=(8, 6))
    ax = sns.heatmap(
        Z,
        cmap=cmap,
        cbar_kws={"label": "Predicted Log of RMSE (fixed effects only)"}
    )

    # Make axes readable (don’t label every grid point)
    xticks = np.linspace(0, n_grid - 1, 6).astype(int)
    yticks = np.linspace(0, n_grid - 1, 6).astype(int)

    ax.set_xticks(xticks)
    ax.set_yticks(yticks)

    ax.set_xticklabels([f"{mi_grid[i]:.2f}" for i in xticks], rotation=0)
    ax.set_yticklabels([f"{moran_grid[i]:.2f}" for i in yticks], rotation=0)

    ax.set_xlabel("MI Shift")
    ax.set_ylabel("Moran Shift")
    # ax.set_title("Joint partial effect of MI Shift and Moran Shift (MixedLM)")

    ax.invert_yaxis()
    plt.tight_layout()


    plt.savefig('models/DGM_raw/evaluations/figures/Joint_partial_effects_RMSE.pdf', bbox_inches='tight')

    plt.show()

    return MI, MORAN, Z
```

```python
MI, MORAN, Z = joint_partial_heatmap(df_analysis_copy, m)
```

```python
def plot_within_state_trends_residualized(df, result, x="moran_shift",
                                          state="src_state", states_to_plot=None,
                                          n_states=6, alpha=0.15):
    d = df.copy()
    d["resid"] = result.resid
    d = d[[state, x, "resid"]].dropna()

    if states_to_plot is None:
        states_to_plot = (
            d[state].value_counts()
            .head(n_states)
            .index
            .tolist()
        )

    fig, ax = plt.subplots(figsize=(8, 5))

    for s in states_to_plot:
        ds = d[d[state] == s]
        ax.scatter(ds[x], ds["resid"], alpha=alpha, s=10)

        xs = ds[x].to_numpy()
        ys = ds["resid"].to_numpy()
        if len(xs) >= 5 and np.nanstd(xs) > 0:
            b1, b0 = np.polyfit(xs, ys, 1)
            xline = np.linspace(xs.min(), xs.max(), 50)
            yline = b1 * xline + b0
            ax.plot(xline, yline, linewidth=2, label=str(s))

    ax.axhline(0, linestyle="--", linewidth=1)
    ax.set_xlabel(x)
    ax.set_ylabel("Residualized outcome")
    ax.set_title(f"Within-state trend lines (residualized): resid vs {x}")
    ax.legend(title=state, bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

# Example:
# plot_within_state_trends_residualized(df, result, x="moran_shift")
```

```python
import statsmodels.api as sm

# y = df_analysis_agg_src_ct['CPC_mean']
y = np.log(df_analysis_agg_src_ct['CPC_mean'] / (1 - df_analysis_agg_src_ct['CPC_mean']))
X_full = df_analysis_agg_src_ct[['moran_shift_mean', 'mi_shift_mean', 'ct_count']]
X_full = feat_scaler.fit_transform(X_full)
X_full = sm.add_constant(X_full)
m_full = sm.OLS(y, X_full).fit()

# reduced (drop MI)
X_no_mi = sm.add_constant(df_analysis_agg_src_ct[['moran_shift_mean', 'ct_count']])
m_no_mi = sm.OLS(y, X_no_mi).fit()

# reduced (drop Moran)
X_no_moran = sm.add_constant(df_analysis_agg_src_ct[['mi_shift_mean', 'ct_count']])
m_no_moran = sm.OLS(y, X_no_moran).fit()

# reduced (drop ct_count)
X_no_count = sm.add_constant(df_analysis_agg_src_ct[['moran_shift_mean','mi_shift_mean',]])
m_no_count = sm.OLS(y, X_no_count).fit()

def partial_r2(r2_full, r2_reduced):
    return (r2_full - r2_reduced) / (1 - r2_reduced)

pr2_mi    = partial_r2(m_full.rsquared, m_no_mi.rsquared)      # unique MI
pr2_moran = partial_r2(m_full.rsquared, m_no_moran.rsquared)   # unique Moran
pr2_count = partial_r2(m_full.rsquared, m_no_count.rsquared)   # unique Moran

pr2_mi, pr2_moran, pr2_count
```

```python
import statsmodels.api as sm

y = df_analysis_agg_src_ct['RMSE_mean']
# y = np.log(df_analysis_agg_src_ct['CPC_mean'] / (1 - df_analysis_agg_src_ct['CPC_mean']))
X_full = df_analysis_agg_src_ct[['moran_shift_mean', 'mi_shift_mean', 'ct_count']]
X_full = feat_scaler.fit_transform(X_full)
X_full = sm.add_constant(X_full)
m_full = sm.OLS(y, X_full).fit()

# reduced (drop MI)
X_no_mi = sm.add_constant(df_analysis_agg_src_ct[['moran_shift_mean', 'ct_count']])
m_no_mi = sm.OLS(y, X_no_mi).fit()

# reduced (drop Moran)
X_no_moran = sm.add_constant(df_analysis_agg_src_ct[['mi_shift_mean', 'ct_count']])
m_no_moran = sm.OLS(y, X_no_moran).fit()

# reduced (drop ct_count)
X_no_count = sm.add_constant(df_analysis_agg_src_ct[['moran_shift_mean','mi_shift_mean',]])
m_no_count = sm.OLS(y, X_no_count).fit()

def partial_r2(r2_full, r2_reduced):
    return (r2_full - r2_reduced) / (1 - r2_reduced)

pr2_mi    = partial_r2(m_full.rsquared, m_no_mi.rsquared)      # unique MI
pr2_moran = partial_r2(m_full.rsquared, m_no_moran.rsquared)   # unique Moran
pr2_count = partial_r2(m_full.rsquared, m_no_count.rsquared)   # unique Moran

pr2_mi, pr2_moran, pr2_count
```

# Task
Calculate a `combined_performance_score` by normalizing `CPC_mean` (higher is better) and `RMSE_mean` (lower is better) to a 0-1 scale and summing the normalized CPC with the normalized (inverted) RMSE. Finally, review the plot and discuss insights.

## Calculate Combined Performance Score

### Subtask:
Normalize 'CPC_mean' (higher is better) and 'RMSE_mean' (lower is better) to a common scale (0 to 1). Then, create a 'combined_performance_score' by summing the normalized CPC and normalized (inverted) RMSE.


**Reasoning**:
First, I will normalize the 'CPC_mean' column in `df_analysis_agg_src` to a 0-1 scale using min-max normalization, as higher CPC is better. This will create a new column named 'normalized_cpc'.



```python
cpc_min = df_analysis_agg_src['CPC_mean'].min()
cpc_max = df_analysis_agg_src['CPC_mean'].max()
df_analysis_agg_src['normalized_cpc'] = (df_analysis_agg_src['CPC_mean'] - cpc_min) / (cpc_max - cpc_min)
display(df_analysis_agg_src.head())
```

**Reasoning**:
Next, I will invert the `RMSE_mean` column in `df_analysis_agg_src` to ensure that lower RMSE values correspond to higher values, which makes it comparable to the CPC score. Then, I will normalize this inverted `RMSE_mean` to a 0-1 range using min-max normalization, storing it in a new column named `normalized_inverted_rmse`.



```python
rmse_max = df_analysis_agg_src['RMSE_mean'].max()
rmse_min = df_analysis_agg_src['RMSE_mean'].min()
df_analysis_agg_src['inverted_rmse'] = rmse_max - df_analysis_agg_src['RMSE_mean']

inv_rmse_min = df_analysis_agg_src['inverted_rmse'].min()
inv_rmse_max = df_analysis_agg_src['inverted_rmse'].max()
df_analysis_agg_src['normalized_inverted_rmse'] = (df_analysis_agg_src['inverted_rmse'] - inv_rmse_min) / (inv_rmse_max - inv_rmse_min)
display(df_analysis_agg_src.head())
```

**Reasoning**:
Now that both CPC and RMSE have been normalized to a comparable scale (0-1), I will calculate the 'combined_performance_score' by summing 'normalized_cpc' and 'normalized_inverted_rmse'. This directly follows the instructions for completing the combined performance score calculation.



```python
df_analysis_agg_src['combined_performance_score'] = df_analysis_agg_src['normalized_cpc'] + df_analysis_agg_src['normalized_inverted_rmse']
display(df_analysis_agg_src.head())
```

**Reasoning**:
Now that the `combined_performance_score` has been calculated, I will create a scatter plot to visualize the relationship between `mi_shift_mean`, `moran_shift_mean`, and the `combined_performance_score`. The size of the dots will reflect the `combined_performance_score`, as requested in the main task.



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

# Create the scatter plot
scatter = plt.scatter(
    df_analysis_agg_src['mi_shift_mean'],
    df_analysis_agg_src['moran_shift_mean'],
    s=df_analysis_agg_src['combined_performance_score'] * 100, # Scale up for better visibility
    c=df_analysis_agg_src['combined_performance_score'],      # Color by the score as well
    cmap='viridis',                                        # Choose a color map
    alpha=0.7,
    edgecolors='w',                                        # White edges for better distinction
    linewidth=0.5
)

# Add a color bar for the combined_performance_score
cbar = plt.colorbar(scatter, label='Combined Performance Score')

plt.xlabel('MI Shift (Mean)')
plt.ylabel('Moran Shift (Mean)')
plt.title('Domain Shifts vs. Combined Performance Score (Size & Color)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Plot Insights

The scatter plot visualizes the relationship between different domain shifts (MI Shift and Moran Shift) and the `combined_performance_score` across various source states. Each point represents a source state, with its position determined by the mean MI Shift and mean Moran Shift, and its size and color reflecting the model's overall performance (a combination of normalized CPC and inverted RMSE).

**Observations:**

*   **Distribution of Shifts:** The plot shows a distribution of source states across different combinations of MI and Moran shifts. This indicates that the domain shift experienced by a model trained on a specific source state, when applied to a target state, varies significantly.
*   **Performance Trends:** Generally, larger and brighter points (indicating higher `combined_performance_score`) tend to cluster in certain regions of the plot. For example, if higher MI Shift and/or Moran Shift values correspond to lower performance, the larger/brighter dots would be in the areas with smaller shift values, and vice versa. This visual correlation can help identify the types of domain shifts that lead to better or worse model performance.
*   **Identifying Optimal/Suboptimal Source Domains:** By examining where the high-scoring (large/bright) and low-scoring (small/dark) points are located, we can infer which combinations of MI and Moran shifts are more conducive to good model transferability. For instance, if large bright dots are concentrated where both MI and Moran shifts are low, it would suggest that models perform best when the source domain is geographically and informationally similar to the target domains.
*   **Potential Trade-offs:** The plot might reveal scenarios where a high shift in one metric (e.g., MI Shift) is acceptable if the other (Moran Shift) is low, or vice versa, influencing the overall performance. This could highlight complex interactions between different types of domain shifts.

**Conclusion:**

This visualization effectively illustrates how the characteristics of the source domain, quantified by MI Shift and Moran Shift, influence the transferability and performance of the model on unseen target domains. It serves as a valuable tool for understanding domain generalization challenges and potentially for selecting optimal source domains for training models that are intended for broad geographic application.
